```python
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    agent="antigravity-preview-05-2026",
    input="Plot the growth of solar energy generation globally and make some slides in HTML.",
    environment="remote",  # Remote linux environment hosted by Google
)

print(interaction.output_text)
```

```python
from google.adk.agents.llm_agent import Agent
from google.adk.models.google_llm import Gemini
from google.adk.tools.google_search_tool import GoogleSearchTool

root_agent = Agent(
    model=Gemini(
        model="gemini-2.5-flash",
        # Enable Interactions API
        use_interactions_api=True,
    ),
    name="interactions_test_agent",
    tools=[
        # Converted Google Search to a function tool
        GoogleSearchTool()
    ],
)
```

```python
from google.adk.runners import Runner
from google.adk.sessions import InMemorySessionService
from google.genai import types

# 1. Initialize the session service
session_service = InMemorySessionService()

# 2. Initialize the runner
runner = Runner(agent=root_agent, session_service=session_service, app_name="test")

# 3. Wrap the prompt in a types.Content object
content = types.Content(
    role="user",
    parts=[
        types.Part.from_text(
            text="Search for the latest global solar energy capacity trends and summarize them in one paragraph."
        )
    ],
)

# 4. Run the root agent asynchronously and print the streamed output
print("Running root_agent...")
async for event in runner.run_async(
    user_id="user_123", session_id="session_001", new_message=content
):
    if event.content and event.content.parts:
        for part in event.content.parts:
            if part.text:
                print(part.text, end="", flush=True)
```
