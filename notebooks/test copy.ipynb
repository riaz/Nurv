```python
# Load environment variables from the .env file (automatically searches parent directories)
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())
```

```python
# Initialize and test the Google GenAI client using the Interactions API
import warnings
from google import genai

# Suppress experimental warnings if desired
warnings.filterwarnings("ignore", category=UserWarning)

client = genai.Client()

print("Initiating interaction with managed agent 'antigravity-preview-05-2026'...")

try:
    interaction = client.interactions.create(
        agent="antigravity-preview-05-2026",
        input="Generate a quick menu concept for a sci-fi cyber-bistro.",
        environment="remote",
    )

    # Extract and display the output
    output_found = False
    if hasattr(interaction, "steps") and interaction.steps:
        for step in reversed(interaction.steps):
            if hasattr(step, "content") and step.content:
                if isinstance(step.content, list):
                    text_parts = []
                    for part in step.content:
                        if hasattr(part, "text"):
                            text_parts.append(part.text)
                        elif isinstance(part, str):
                            text_parts.append(part)
                    if text_parts:
                        print("\nOutput:")
                        print("\n".join(text_parts))
                        output_found = True
                        break
                elif hasattr(step.content, "text"):
                    print("\nOutput:")
                    print(step.content.text)
                    output_found = True
                    break

    if (
        not output_found
        and hasattr(interaction, "output_text")
        and interaction.output_text
    ):
        print("\nOutput:")
        print(interaction.output_text)
        output_found = True

    if not output_found:
        print(interaction)

except Exception as e:
    print(f"\nManaged Agent API returned an error: {e}")
    print(
        "Make sure 'environment' key or other required parameters are provided if necessary."
    )
```
