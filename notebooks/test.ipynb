```python
# 1. Load environment variables from the .env file (automatically searches parent directories)
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())
```

```python
# 2. Import the Google GenAI SDK and A2A Agents
import sys
import os
import json

sys.path.append(
    os.path.abspath("..")
)  # Add parent directory to path to find a2a_collaboration

from google import genai
from a2a_collaboration import A2AOrchestrator

client = genai.Client()
orchestrator = A2AOrchestrator(client)
```

```python
# 3. Run Tokyo Query (Raining in Tokyo -> Applies 1500 JPY/10 USD taxi surcharge)
tokyo_query = "I want to eat a 3000 JPY lunch in Tokyo. If it's raining, I need a taxi. What is the total cost in USD?"
tokyo_result = orchestrator.process(tokyo_query, city="Tokyo")
print("\n--- TOKYO RESULT ---")
print(json.dumps(tokyo_result, indent=2))
```

```python
# 4. Run New York Query (Sunny in NY -> No taxi surcharge)
ny_query = "I want to eat a 50 USD dinner in New York. If it's raining, I need a taxi. What is the total cost in USD?"
ny_result = orchestrator.process(ny_query, city="New York")
print("\n--- NEW YORK RESULT ---")
print(json.dumps(ny_result, indent=2))
```
